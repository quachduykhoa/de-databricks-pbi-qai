```python
path_shops = "/Volumes/workspace/lab_tuan1/raw/shops.csv"

schema_ddl = """
  shop_id        STRING,
  shop_code      STRING,
  shop_name      STRING,
  address        STRING,
  region         STRING,
  opened_date    DATE,
  status         STRING
"""

df_shops = spark.read.csv(path_shops, header=True, schema=schema_ddl)
display(df_shops)
df_shops.printSchema()
```

```python
df_shops.createOrReplaceTempView("v_shops")

```

```python
%sql
-- 1) đếm tổng số shop
SELECT count(*) FROM v_shops;

-- 2) đếm số shop đang ACTIVE
SELECT count(*) FROM v_shops WHERE status = "ACTIVE"
```
